# Face Mask Detection (Local)

এই notebook আপনার PC-এর `data/with_mask` এবং `data/without_mask` folder ব্যবহার করে model train করবে।

```python
from argparse import Namespace
from pathlib import Path

from train_model import train

PROJECT_DIR = Path.cwd()
if not (PROJECT_DIR / 'data').is_dir():
    raise FileNotFoundError('Project root-এ notebook চালান, যেখানে data folder আছে।')

args = Namespace(
    data_dir=PROJECT_DIR / 'data',
    output=PROJECT_DIR / 'models' / 'face_mask_model.pt',
    epochs=5,
    batch_size=32,
    learning_rate=1e-3,
    seed=2,
)
print(f'Dataset: {args.data_dir}')
print(f'Model output: {args.output}')
```

```python
train(args)
```

```python
from PIL import Image
from app import load_model, predict

model, class_names = load_model()
sample_path = next((args.data_dir / 'with_mask').glob('*.jpg'))
label, confidence = predict(model, Image.open(sample_path), class_names)
print(f'Prediction: {label} ({confidence:.2%})')
```
